```python
from textblob import TextBlob
from newspaper import Article
import nltk
nltk.download('punkt_tab')
```

```python
url  = 'https://www.goodnewsnetwork.org/'
article = Article(url)
article.download()
article.parse()
article.nlp()
```

```python
text = article.summary
text
```

```python
blob = TextBlob(text)
sentiment = blob.sentiment.polarity
sentiment
```
